```python
import os
import csv
```

```python
def getDataFileNames(folderPath):
    fileNames = []
    
    try:
        if not os.path.isdir(folderPath):
            print("Error: Folder path doesn't exist. Please check and try again.")
            return

        for file in os.listdir(folderPath):
            if os.path.isfile(os.path.join(folderPath, file)):
                fileNames.append(file)

        return fileNames
        
    except Exception as e:
        print("Error:"+e)
        return

folderPath = r"D:\DATA Projetcs\Amazon Sales\Dataset"
dataFileNames = getDataFileNames(folderPath)
print(dataFileNames)
```

```python
def mergeData(dataFileNames, folderPath):
    try:
        allRows = []
        headerWritten = False
        
        for fileName in dataFileNames:
            try:
                filePath = os.path.join(folderPath, fileName)
    
                with open(filePath, 'r', newline='', encoding="utf-8") as infile:
                    reader = csv.reader(infile)
                    currentHeader = next(reader, None)
    
                    if not headerWritten and currentHeader:
                        allRows.append(currentHeader)
                        headerWritten = True
                    elif headerWritten and currentHeader:
                        pass
    
                    for row in reader:
                        allRows.append(row)
                        
            except Exception as e:
                print(f"Error while reading the data file '{fileName}' -> {e}")
                return

            if allRows:
                try:
                    with open("AmazonSalesDataSet1.csv", 'w', newline='', encoding='utf-8') as outfile:
                        writer = csv.writer(outfile)
                        writer.writerows(allRows)
                    print(f"Data merged successfully in AmazonSalesDataSet - for {fileName}")

                except Exception as e:
                    print(f"Error occured while writing data in AmazonSalesDataSet -> {e}")
                    return
            else:
                print(f"No data found in {fileName}")    
                return

    except Exception as e:
        print(f"Unexpected Error occured -> {e}")
        
mergeData(dataFileNames, folderPath)        
```

```python
def cleaningData(dataFile):
    cleanedRows = []

    try:
        with open(dataFile, 'r', newline='', encoding='utf-8') as infile:
            reader = csv.reader(infile)
            header = next(reader, None)

            if header:
                cleanedRows.append(header)
            else:
                print("Error -> File seems to be empty. Please check and try again.")
                return

            for row in reader:
                cleanedRow = []
                for i, value in enumerate(row):
                    cleanedValue = value.strip()
                    if not cleanedValue:
                        cleanedValue = '0'

                    columnName = header[i] if header and i < len(header) else str(i)
                    if columnName in ['discount_price','actual_price'] or str(i) in ['discount_price','actual_price']:
                        cleanedValue = ''.join(filter(str.isdigit, cleanedValue))

                    cleanedRow.append(cleanedValue)
                cleanedRows.append(cleanedRow)
    except Exception as e:
        print(f"Unexpected Error occured while reading AmazonSalesDataSet1 -> {e}")

    try:
        with open("cleanedAmazonSalesDataSet.csv", 'w', newline='', encoding='utf-8') as outfile:
            writer = csv.writer(outfile)
            writer.writerows(cleanedRows)
        print("Cleaned Data written to cleanedAmazonSalesDataSet.csv")

    except Exception as e:
        print(f"Unexpected Error occured while writing in cleanedAmazonSalesDataSet.csv-> {e}")


dataFile = r"D:\DATA Projetcs\Amazon Sales\AmazonSalesDataSet1.csv"
cleaningData(dataFile)
```
