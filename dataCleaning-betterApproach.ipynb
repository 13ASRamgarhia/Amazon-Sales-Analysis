```python
import os
import pandas as pd
import numpy as np
import csv
```

```python
def getDataFileNames(folderPath):
    fileNames = []
    
    try:
        if not os.path.isdir(folderPath):
            print("Error: Folder path doesn't exist. Please check and try again.")
            return

        for file in os.listdir(folderPath):
            if os.path.isfile(os.path.join(folderPath, file)):
                fileNames.append(file)

        return fileNames
        
    except Exception as e:
        print("Error:"+e)
        return

folderPath = r"D:\DATA Projetcs\Amazon Sales\amazon\Dataset"
dataFileNames = getDataFileNames(folderPath)
print(dataFileNames)
```

```python
def dataCleaning(df, dataFileName):
    try:
        dfCopy = df.copy()
        colsToModify = ['ratings', 'no_of_ratings', 'discount_price', 'actual_price']

        #Dropping unwanted columns
        dfCopy = dfCopy.drop(columns=["name", "image", "link"], axis=1)

        dfCopy['main_category'] = dfCopy['main_category'].astype(str).str.title()
        
        for col in colsToModify:
            #Replacing all NULL values with '0'
            dfCopy[col] = dfCopy[col].fillna(0)

            #Removing non-digit characters
            dfCopy[col] = dfCopy[col].astype(str).str.replace(r'[^0-9.]', '', regex=True)

        return dfCopy
            
    except Exception as e:
        print(f"Unexpected error while cleaning data -> {e}")
        return


try:
    finalDF = pd.DataFrame()
    
    for dataFileName in dataFileNames:
        df = pd.read_csv(os.path.join(folderPath, dataFileName))
        cleanedDF = dataCleaning(df, dataFileName)

        if cleanedDF is not None:
            print(f"-"*10)
            print(f"Data cleaned for '{dataFileName}' successfully")
            finalDF = pd.concat([finalDF, cleanedDF], ignore_index=True)
            print(f"Data added for '{dataFileName}'")
        else:
            print(f"ERROR -> Function returned None for '{dataFileName}'")
            break
        
    finalDF.to_csv("cleanedDataset7.csv", index=False)
    print(f"Cleaned Dataset created successfully")
        
except Exception as e:
    print(f"Unexpected error -> {e}")
```
